```python
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110,180,240,99,220])
```

```python
fig, (ax1, ax2) = plt.subplots(ncols = 2,
                               nrows = 1,
                               figsize = (7,3))

ax1.scatter(x1,y1, color= 'indigo')
ax2.bar(x2,y2, color = 'pink', edgecolor = 'black')

font1 = {'family': 'serif', 'color': 'grey', 'size':10}
ax1.set_xlabel('X Axis', fontdict=font1)

font2 = {'family': 'serif', 'color': 'indigo', 'size':10}
ax2.set_xlabel('New X Axis', fontdict=font2)

ax1.yaxis.set_major_locator(plt.NullLocator())
ax2.yaxis.set_major_locator(plt.NullLocator())

ax1.set_title('Scatterplot', fontdict=font1, loc='left')
ax2.set_title('Bar', fontdict=font2, loc='left')
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4),
                                    layout = "constrained")
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

someaxes["ax1"].set_xlabel('A Big Label', fontsize = 18)
someaxes["ax2"].set_xlabel('Another Label', fontsize = 18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
```

```python
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110,180,240,99,220])
y2 = np.array([170,100,90,120,50])

fig, ax = plt.subplots(figsize=(7,3))

ax.bar(x,y1, color = 'indigo')
ax.plot(x,y2, color="red")
```

```python
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x,y2, color = "red")

ax.errorbar(x,y2,
           yerr = y2_sd,
           fmt = "none",
           ecolor = 'indigo',
           elinewidth = 4,
           capsize = 6,
           capthick = 4,
           errorevery = 2)
```
