```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
sns.set_style('dark')

sns.lineplot(data = tips,
             x = 'total_bill',
             y = 'tip')
```

```python
tipgraph = sns.lineplot(data = tips,
                        x = 'total_bill',
                        y = 'tip')

tipgraph.set(title = 'Tips vs. Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
fig = plt.subplots(figsize=(10, 3))

tipgraph = sns.lineplot(data=tips,
                        x = 'total_bill',
                        y = 'tip',
                        color = 'hotpink',
                        linestyle = '--',
                        linewidth = 3,
                        marker = 'o',
                        markerfacecolor = 'indigo')

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
tipgraph = sns.scatterplot(data=tips,
                        x = 'total_bill',
                        y = 'tip',
                        style = 'time',
                        hue = 'day',
                        palette = ['purple', 'hotpink', 'deepskyblue', 'yellowgreen'])
```

```python
sns.pairplot(data = tips,
             hue = 'day')
```

```python
daysplot = sns.relplot(data = tips,
                       x = 'total_bill',
                       y = 'tip',
                       hue = 'sex',
                       col = 'day',
                       kind = 'scatter',
                       col_wrap = 2)
```

plotting total_bill vs tip for different days with color based on sex column

```python
import plotly.graph_objects as go

x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
graph = go.Figure()

graph.add_trace(go.Bar(x=x1, y=y1))

graph.update_layout(title = "Pirate Scores",
                    xaxis_title = "Pirtes",
                    yaxis_title = "Score")

graph.show()
```

```python
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
                on_bad_lines = 'skip')

df
```

```python
text = " ".join(each for each in df.quote)

wordcloud = WordCloud(background_color = 'white',
                        colormap = 'inferno').generate(text)

fig, ax = plt.subplots(figsize=(7,3))
ax.imshow(wordcloud)
ax.axis('off')
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = set(["apple", "banna", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])

diagram = venn2([A,B],
            set_labels = ('Fruits', 'Pires'),
            set_colors = ('blue', 'red'),
            alpha = 0.5)

diagram.get_label_by_id("10").set_text("\n".join(A-B)) #left
diagram.get_label_by_id("11").set_text("\n".join(A&B)) #middle
diagram.get_label_by_id("01").set_text("\n".join(B-A)) #right
```
