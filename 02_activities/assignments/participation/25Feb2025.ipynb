```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.hist(y)
```

```python
font1 = {'family': 'sans-serif', 'color': 'blue', 'size':20}
font2 = {'family': 'monospace', 'color': 'green', 'size':14}

fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)

ax.set_title('Total growth over time', fontdict=font1, loc='left')
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)

plt.show()
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y,
           marker = '*',
           color = 'indigo')
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y,
        marker = '*',
        color = 'indigo',
        linestyle = '--',
        linewidth = 2,
        markersize = 12,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#000000')

#ax.grid(axis = 'x')
ax.grid()
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y,
        marker = '*',
        color = 'indigo',
        linestyle = '--',
        linewidth = 1,
        markersize = 12,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#000000')

#ax.grid(axis = 'x')
ax.grid(axis= 'x',
        color='green', linestyle='dotted', linewidth=1)
```

```python
import folium

# Make an empty map
m = folium.Map(location=[20,0], tiles="OpenStreetMap", zoom_start=2)

# Show the map
m
```

```python
# Make a data frame with dots to show on the map
data = pd.DataFrame({
   'lon':[-58, 2, 145, 30.32, -4.03, -73.57, 36.82, -38.5],
   'lat':[-34, 49, -38, 59.93, 5.33, 45.52, -1.29, -12.97],
   'name':['Buenos Aires', 'Paris', 'melbourne', 'St Petersbourg', 'Abidjan', 'Montreal', 'Nairobi', 'Salvador'],
   'value':[10, 12, 40, 70, 23, 43, 100, 43]
}, dtype=str)

data
```

```python
for idx, row in data.iterrows():
    folium.CircleMarker(
        location=[row['lat'], row['lon']],
        radius=float(row['value']),
        popup=f"{row['name']} ({row['value']})",
        color="green", #The border color of the circle
        fill=True, #Whether to fill the circle’s interior
        fill_color="red", #The color used to fill the circle
        fill_opacity=0.2, #The transparency of the fill (0 to 1)
        weight = 0 #The thickness of the circle’s border in pixels
    ).add_to(m)

m
```
